```python
def get_number(message):
#asking the user to enter a positive whole number.

    while True:
        try:
            number = int(input(message))

            if number > 0:
                return number
#the number should be greater than zero

            print("Enter a number greater than zero.")

        except ValueError:
#if the user enters letters or an invalid value
            print("Please enter a valid whole number.")


def get_quantity(message):
#asking the user to enter a zero or a positive whole number.

    while True:
        try:
            number = int(input(message))

            if number >= 0:
                return number

            print("Enter zero or a positive number.")

        except ValueError:
            print("Please enter a valid whole number.")


def get_price(message):
#asking the user to enter a valid price

    while True:
        try:
            price = float(input(message))

            if price > 0:
                return price

            print("Price must be greater than zero.")

        except ValueError:
            print("Please enter a valid price.")


def read_products():
#get all the saved products from the file

    products = []

    try:
        with open("products.txt", "r") as file:
            for line in file:
                parts = line.strip().split("|")

                if len(parts) == 5:
#storing product details in a dictionary
                    product = {
                        "id": int(parts[0]),
                        "name": parts[1],
                        "category": parts[2],
                        "price": float(parts[3]),
                        "stock": int(parts[4])}

                    products.append(product)

    except FileNotFoundError:
        print("Product file not found.")

    except PermissionError:
        print("Permission denied while reading products.")

    except (ValueError, OSError) as error:
        print("Product file error:", error)
#returning the product list
    return products


def save_products(products):
#saving all products in the file

    try:
        with open("products.txt", "w") as file:
            for product in products:
                file.write(
                    f"{product['id']}|{product['name']}|"
                    f"{product['category']}|{product['price']}|"
                    f"{product['stock']}\n")

        return True

    except (PermissionError, OSError) as error:
        print("Could not save products:", error)
        return False


def next_id(records):
#finding the next ID for a new record

    highest = 0

    for record in records:
        if record["id"] > highest:
            highest = record["id"]
#adding 1 to the highest ID
    return highest + 1


def find_product(products, product_id):
#search for a product using its ID.

    for product in products:
        if product["id"] == product_id:
            return product
            
#returning None if the product is not found

    return None
```

```python
def add_product():
    print("\n--- Add New Product ---")

    # Getting product details from the user
    name = input("Enter product name: ").strip()
    category = input("Enter product category: ").strip()

    # Checking that name and category are not empty
    if name == "" or category == "":
        print("Name and category cannot be empty.")
        return

    # Making sure the user does not enter |
    if "|" in name or "|" in category:
        print("Please do not use the | symbol.")
        return

    # Reading the products already saved in the file
    products = read_products()

    # Getting price and available stock
    price = get_price("Enter product price: Rs. ")
    stock = get_quantity("Enter available stock: ")

    # Creating a new product with its details
    product = {
        "id": next_id(products),
        "name": name,
        "category": category,
        "price": price,
        "stock": stock
    }

    # Adding the new product to the list
    products.append(product)

    # Saving the updated product list
    if save_products(products):
        print("Product added successfully.")
        print("Product ID:", product["id"])
```

```python
def view_products():
    print("\n--- Available Products ---")

    # Getting all products from the file
    products = read_products()

    # Checking if there are any products to display
    if len(products) == 0:
        print("No products available.")
        return

    # Displaying the details of each product
    for product in products:
        print("--------------------------------")
        print("Product ID:", product["id"])
        print("Name:", product["name"])
        print("Category:", product["category"])
        print("Price: Rs.", product["price"])
        print("Available Stock:", product["stock"])

    print("--------------------------------")
```

```python
def search_product():
    print("\n--- Search Product ---")

    # Asking the user what product they want to search for
    keyword = input("Enter product name or category: ").strip().lower()

    # Checking if the search box is empty
    if keyword == "":
        print("Please enter a search term.")
        return

    # Getting the saved products
    products = read_products()
    found = False

    # Checking each product to find a match
    for product in products:
        if (keyword in product["name"].lower()
                or keyword in product["category"].lower()):

            # Showing the details of the matching product
            print("Product ID:", product["id"])
            print("Name:", product["name"])
            print("Category:", product["category"])
            print("Price: Rs.", product["price"])
            print("Stock:", product["stock"])

            found = True

    # Showing a message if no matching product is found
    if found == False:
        print("Product not found.")
```

```python
def update_product():
    print("\n--- Update Product ---")

    # Getting the list of saved products
    products = read_products()
    product_id = get_number("Enter product ID: ")

    # Finding the product using its ID
    product = find_product(products, product_id)

    if product is None:
        print("Product not found.")
        return

    # Showing the options for updating a product
    print("1. Update Price")
    print("2. Update Stock")

    choice = input("Choose an option: ")

    # Updating the product price
    if choice == "1":
        product["price"] = get_price("Enter new price: Rs. ")

    # Updating the available stock
    elif choice == "2":
        product["stock"] = get_quantity("Enter new stock: ")

    else:
        print("Invalid choice.")
        return

    # Saving the updated product details
    if save_products(products):
        print("Product updated successfully.")
```

```python
def delete_product():
    print("\n--- Delete Product ---")

    # Getting all the saved products
    products = read_products()
    product_id = get_number("Enter product ID: ")

    # Finding the product that the user wants to delete
    product = find_product(products, product_id)

    if product is None:
        print("Product not found.")
        return

    # Showing which product is selected
    print("Selected Product:", product["name"])

    # Asking the user before deleting the product
    confirm = input("Do you want to delete it? (yes/no): ").lower()

    if confirm == "yes":
        new_products = []

        # Keeping all products except the one being deleted
        for item in products:
            if item["id"] != product_id:
                new_products.append(item)

        # Saving the list without the deleted product
        if save_products(new_products):
            print("Product deleted successfully.")

    else:
        # If the user does not confirm, nothing is deleted
        print("Deletion cancelled.")
```

```python
def register_customer():
    print("\n--- Register Customer ---")

    # Getting customer details from the user
    name = input("Enter customer name: ").strip()
    phone = input("Enter phone number: ").strip()

    # Checking that both fields are filled
    if name == "" or phone == "":
        print("Customer name and phone are required.")
        return

    # Making sure | is not used in the details
    if "|" in name or "|" in phone:
        print("Please do not use the | symbol.")
        return

    # Creating an empty list to store customer details
    customers = []

    try:
        # Reading the customers already saved in the file
        with open("customers.txt", "r") as file:
            for line in file:
                parts = line.strip().split("|")

                # Adding customer details to the list
                if len(parts) == 3:
                    customers.append({
                        "id": int(parts[0]),
                        "name": parts[1],
                        "phone": parts[2]
                    })

    except FileNotFoundError:
        # If the file does not exist, continue with an empty list
        pass

    except (PermissionError, ValueError, OSError) as error:
        print("Customer file error:", error)
        return

    # Checking if the customer is already registered
    for customer in customers:
        if customer["phone"] == phone:
            print("Customer already registered.")
            return

    # Giving the new customer a unique ID
    customer_id = next_id(customers)

    try:
        # Saving the new customer details in the file
        with open("customers.txt", "a") as file:
            file.write(f"{customer_id}|{name}|{phone}\n")

        print("Customer registered successfully.")
        print("Customer ID:", customer_id)

    except (PermissionError, OSError) as error:
        print("Could not save customer:", error)
```

```python
def view_customers():
    print("\n--- Customer List ---")

    try:
        # Opening the file and reading all customer records
        with open("customers.txt", "r") as file:
            customers = file.readlines()

        # Checking if there are any customers in the file
        if len(customers) == 0:
            print("No customers registered.")
            return

        # Displaying all registered customers
        for customer in customers:
            print(customer.strip())

    except FileNotFoundError:
        print("Customer file not found.")

    except (PermissionError, OSError) as error:
        print("Customer file error:", error)
```

```python
def place_order():
    print("\n--- Place Customer Order ---")

    # Getting all available products
    products = read_products()

    if len(products) == 0:
        print("No products available.")
        return

    # Getting customer details
    name = input("Enter customer name: ").strip()
    phone = input("Enter customer phone: ").strip()

    # Checking that customer details are not empty
    if name == "" or phone == "":
        print("Customer details are required.")
        return

    # Making sure | is not used in customer details
    if "|" in name or "|" in phone:
        print("Please do not use the | symbol.")
        return

    # Creating an empty list for selected products
    items = []
    subtotal = 0

    # Showing all available products
    view_products()

    # Allowing the customer to select multiple products
    while True:
        try:
            product_id = int(input("Enter product ID (0 to finish): "))

            # Ending product selection when the user enters 0
            if product_id == 0:
                break

            if product_id < 0:
                print("Enter a positive product ID.")
                continue

        except ValueError:
            print("Please enter a valid product ID.")
            continue

        # Finding the selected product
        product = find_product(products, product_id)

        if product is None:
            print("Product not found.")
            continue

        # Asking how many items the customer wants
        quantity = get_number("Enter quantity: ")

        # Checking if this product was already added to the order
        already_added = 0

        for item in items:
            if item["id"] == product_id:
                already_added += item["quantity"]

        # Making sure enough stock is available
        if already_added + quantity > product["stock"]:
            print("Not enough stock available.")
            continue

        # Calculating the total price of this product
        item_total = product["price"] * quantity

        # Adding the selected product to the order
        items.append({
            "id": product_id,
            "name": product["name"],
            "quantity": quantity,
            "total": item_total
        })

        # Adding the product price to the subtotal
        subtotal += item_total

        print("Product added to order.")

    # Checking if the customer selected any products
    if len(items) == 0:
        print("No products selected. Order cancelled.")
        return

    # Giving a 5% discount on orders of Rs. 5000 or more
    discount = 0

    if subtotal >= 5000:
        discount = subtotal * 0.05

    # Calculating the final bill after discount
    final_total = subtotal - discount

    # Displaying the customer bill
    print("\n--- Customer Bill ---")
    print("Customer:", name)
    print("Subtotal: Rs.", round(subtotal, 2))
    print("Discount: Rs.", round(discount, 2))
    print("Final Total: Rs.", round(final_total, 2))

    # Asking the customer to confirm the order
    confirm = input("Confirm order? (yes/no): ").lower()

    if confirm != "yes":
        print("Order cancelled.")
        return

    # Reducing the stock of the selected products
    for item in items:
        product = find_product(products, item["id"])
        product["stock"] -= item["quantity"]

    # Saving the updated stock in the products file
    if save_products(products) == False:
        print("Order cancelled because stock could not be saved.")
        return

    # Saving the order details in the orders file
    try:
        with open("orders.txt", "a") as file:
            file.write(
                f"{name}|{phone}|{round(subtotal, 2)}|"
                f"{round(discount, 2)}|{round(final_total, 2)}\n"
            )

        print("Order placed successfully.")

    except (PermissionError, OSError) as error:
        print("Could not save order:", error)

        # Adding the stock back if the order could not be saved
        for item in items:
            product = find_product(products, item["id"])
            product["stock"] += item["quantity"]

        # Saving the restored stock
        save_products(products)
```

```python
def view_orders():
    print("\n--- Order History ---")

    try:
        # Opening the file and reading all saved orders
        with open("orders.txt", "r") as file:
            orders = file.readlines()

        # Checking if there are any orders in the file
        if len(orders) == 0:
            print("No orders available.")
            return

        # Displaying each order one by one
        for order in orders:
            # Separating the order details using |
            parts = order.strip().split("|")

            if len(parts) == 5:
                print("--------------------------------")
                print("Customer:", parts[0])
                print("Phone:", parts[1])
                print("Subtotal: Rs.", parts[2])
                print("Discount: Rs.", parts[3])
                print("Final Total: Rs.", parts[4])

    except FileNotFoundError:
        print("Order file not found.")

    except (PermissionError, OSError) as error:
        print("Order file error:", error)
```

```python
def inventory_report():
    print("\n--- Inventory Report ---")

    # Getting all products from the file
    products = read_products()

    # Checking if there are any products available
    if len(products) == 0:
        print("No inventory available.")
        return

    # Starting stock and value from zero
    total_units = 0
    total_value = 0

    # Calculating total stock and inventory value
    for product in products:
        total_units += product["stock"]
        total_value += product["price"] * product["stock"]

    # Displaying the inventory summary
    print("Total Products:", len(products))
    print("Total Stock Units:", total_units)
    print("Total Inventory Value: Rs.", round(total_value, 2))

    print("\nLow Stock Products:")

    found = False

    # Checking which products have 5 or fewer items left
    for product in products:
        if product["stock"] <= 5:
            print(product["name"], "| Stock:", product["stock"])
            found = True

    # Showing a message if no products have low stock
    if found == False:
        print("No low-stock products.")
```

```python
def sales_report():
    print("\n--- Sales Report ---")

    try:
        # Opening the file and reading all saved orders
        with open("orders.txt", "r") as file:
            orders = file.readlines()

        # Starting sales and order count from zero
        total_sales = 0
        total_orders = 0

        # Going through each order to calculate total sales
        for order in orders:
            parts = order.strip().split("|")

            # Checking if the order has all 5 details
            if len(parts) == 5:
                total_sales += float(parts[4])
                total_orders += 1

        # Displaying the total number of orders and sales
        print("Total Orders:", total_orders)
        print("Total Sales: Rs.", round(total_sales, 2))

    except FileNotFoundError:
        print("Order file not found.")

    except (PermissionError, ValueError, OSError) as error:
        # Showing an error if something goes wrong
        print("Sales report error:", error)
```

```python
def main_menu():

    # Keeping the menu running until the user chooses to exit
    while True:

        print("\n======================================")
        print(" FLOWER AND GIFT SHOP MANAGEMENT")
        print("======================================")

        # Showing all the available options
        print("1. Add Product")
        print("2. View Products")
        print("3. Search Product")
        print("4. Update Product")
        print("5. Delete Product")
        print("6. Register Customer")
        print("7. View Customers")
        print("8. Place Order")
        print("9. View Orders")
        print("10. Inventory Report")
        print("11. Sales Report")
        print("0. Exit")

        # Asking the user to select an option
        choice = input("\nEnter your choice: ").strip()

        # Calling the function according to the user's choice
        if choice == "1":
            add_product()

        elif choice == "2":
            view_products()

        elif choice == "3":
            search_product()

        elif choice == "4":
            update_product()

        elif choice == "5":
            delete_product()

        elif choice == "6":
            register_customer()

        elif choice == "7":
            view_customers()

        elif choice == "8":
            place_order()

        elif choice == "9":
            view_orders()

        elif choice == "10":
            inventory_report()

        elif choice == "11":
            sales_report()

        # Exiting the program
        elif choice == "0":
            print("Thank you for using the Flower and Gift Shop Portal.")
            break

        # Showing a message if the user enters the wrong option
        else:
            print("Invalid choice. Please try again.")
```
